```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import precision_score, recall_score, f1_score

# ۱. ساختن یک دیتاست فرضی برای امتیاز کاربران به فیلم‌ها
# 0 یعنی کاربر فیلم رو ندیده
data = {
    'User1': [5, 4, 0, 1],
    'User2': [4, 5, 2, 0],
    'User3': [1, 0, 5, 4]
}
df = pd.DataFrame(data, index=['MovieA', 'MovieB', 'MovieC', 'MovieD'])

# ۲. محاسبه شباهت کسینوسی بین کاربران (ماتریس شباهت)
user_similarity = cosine_similarity(df.T)
sim_df = pd.DataFrame(user_similarity, index=df.columns, columns=df.columns)
print("ماتریس شباهت کسینوسی کاربران:")
print(sim_df.round(2))

# ۳. پیش‌بینی امتیاز برای کاربر هدف (User1) روی فیلمی که ندیده (MovieC)
target_user = 'User1'
item_to_predict = 'MovieC'
similar_users = df.columns[df.loc[item_to_predict] > 0]

numerator = 0
denominator = 0
target_user_index = df.columns.get_loc(target_user)

for user in similar_users:
    if user != target_user:
        user_index = df.columns.get_loc(user)
        similarity = user_similarity[target_user_index][user_index]
        rating = df.loc[item_to_predict, user]

        numerator += similarity * rating
        denominator += similarity

predicted_rating = numerator / denominator if denominator != 0 else 0
print(f"\nپیش‌بینی امتیاز {target_user} برای {item_to_predict}: {predicted_rating:.2f}")
```
