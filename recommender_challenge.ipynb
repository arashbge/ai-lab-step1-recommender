```python
import pandas as pd

# دیتاست اولیه فیلم‌ها
movies_df = pd.DataFrame({
    'movieId': [1, 2, 3, 4, 5],
    'title': ['Action Movie 1', 'Comedy 1', 'Action 2', 'Drama 1', 'Sci-Fi 1'],
    'genres': ['Action', 'Comedy', 'Action', 'Drama', 'Sci-Fi']
})

# دیتاست اولیه امتیاز کاربران (کاربر 1 به اکشن‌ها علاقه داره)
ratings_df = pd.DataFrame({
    'userId': [1, 1, 2, 2],
    'movieId': [1, 3, 2, 4],
    'rating': [5, 4, 5, 4]
})
print("دیتاست اولیه آماده شد.")
```

```python
class MovieRecommender:
    def __init__(self, ratings_df, movies_df):
        self.ratings = ratings_df
        self.movies = movies_df

    def recommend_for_user(self, user_id):
        user_ratings = self.ratings[self.ratings["userId"] == user_id]
        liked_movies = user_ratings[user_ratings["rating"] >= 4]["movieId"]
        liked_genres = self.movies[self.movies["movieId"].isin(liked_movies)]["genres"].str.split("|")
        genre_counts = pd.Series([g for sublist in liked_genres for g in sublist]).value_counts()

        if len(genre_counts) == 0: return "هیچ پیشنهادی وجود ندارد"

        fav_genre = genre_counts.index[0]
        candidates = self.movies[
            self.movies["genres"].str.contains(fav_genre, na=False)
            & ~self.movies["movieId"].isin(liked_movies)
        ]
        return candidates['title'].tolist()
```

```python
system = MovieRecommender(ratings_df, movies_df)
print("پیشنهاد اولیه برای کاربر 1 (که عاشق اکشن است):", system.recommend_for_user(1))
```

```python
# حالا شبیه‌سازی می‌کنیم که زمان گذشته و سلیقه کاربر عوض شده است!
# کاربر 1 حالا یک فیلم کمدی می‌بیند و به آن امتیاز 5 می‌دهد (محیط تغییر کرد)
new_rating = pd.DataFrame({'userId': [1], 'movieId': [2], 'rating': [5]})
system.ratings = pd.concat([system.ratings, new_rating], ignore_index=True)
print("محیط تغییر کرد: سلیقه کاربر 1 به سمت کمدی رفت.")
```

```python
print("پیشنهاد جدید برای کاربر 1 پس از تغییر سلیقه:", system.recommend_for_user(1))
```
